```python
import pandas as pd
```

```python
df = pd.read_csv('./raw/hate_crime.csv')
```

```python
df.info()
```

```python
df.shape
```

```python
print("\nMissing values per column:\n", df.isna().sum())
```

```python
drop_cols = [
    "incident_id",
    "ori",
    "pug_agency_name",
    "pub_agency_unit",
    "division_name",
    "region_name",
    "population_group_code",
    "population_group_description"
]

df = df.drop(columns=drop_cols, errors="ignore")

df["incident_date"] = pd.to_datetime(df["incident_date"], errors="coerce")
```

```python
df = df.replace(["", "NULL", "NaN", "NA"], pd.NA)
```

```python
df = df.reset_index(drop=True)
```

```python
print("\nMissing values per column:\n", df.isna().sum())
```

```python
df.to_csv("./preproccesed/hate_crime_preproccesed.csv", index=False)
```
